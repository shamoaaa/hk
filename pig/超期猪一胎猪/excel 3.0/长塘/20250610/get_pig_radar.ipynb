```python
import pandas as pd

# 读取 Excel 文件路径
file_path = r'E:\5.毫米波雷达\code\一步到位\excel 3.0\长塘\长塘场查情配种记录.xlsx'

# 选择要排序和删除重复值的列名称（例如 'number_column'）
number_column = '栏位'

# 读取 Excel 文件
df = pd.read_excel(file_path)

# 初始化一个新的列表来存储处理后的数据
new_data = []

# 将数据排序
df_sorted = df.sort_values(by=number_column)

# # 查看排序后的数据
# print("Sorted DataFrame:")
# print(df_sorted)

# 获取数字列的值
numbers = df_sorted[number_column].values

# 遍历数字列，检测缺失的部分并插入行
expected_value = numbers[0]
for index, value in enumerate(numbers):
    while expected_value < value:
        # 创建一个DataFrame用于存放缺失行
        missing_row_df = pd.DataFrame({number_column: [expected_value]}, index=[index])
        new_data.append(missing_row_df)
        expected_value += 1
    new_data.append(df_sorted.iloc[[index]])  # 使用双重括号以带上原来DataFrame的属性
    expected_value += 1

# 拼接所有数据
new_df = pd.concat(new_data).reset_index(drop=True)

# # 查看处理后的数据
# print("DataFrame with Missing Rows Inserted:")
# print(new_df)

# 如果需要保存结果到新的 Excel 文件中
output_file_path = r'E:\5.毫米波雷达\code\一步到位\excel 3.0\长塘\长塘场查情配种记录-填充.xlsx'
new_df.to_excel(output_file_path, index=False)
```

```python
def string_to_combined_hex_to_decimal(input_string):
    if len(input_string) == 5:
        return input_string
        
    # 初始化空字符串用于保存十六进制表示
    combined_hex = ""

    for char in input_string:
        # 将字符转换为两位的十六进制表示并添加到combined_hex中
        combined_hex += format(ord(char), "02x")

    # 将完整的十六进制字符串转换为十进制数值
    decimal_value = int(combined_hex, 16)

    return str(decimal_value)
```

```python
def get_algo_result(directory, col_index, algo_result):
    # 获取目录下的所有项
    all_items = os.listdir(directory)

    # 过滤出所有的 CSV 文件
    csv_files = [item for item in all_items if item.endswith('.csv') and os.path.isfile(os.path.join(directory, item))]
    csv_files = sorted(csv_files)

    # 遍历每个 CSV 文件并将第二列拼接到最终的 DataFrame 中
    for x, csv_file in enumerate(csv_files):

        # 文件的完整路径
        file_path = os.path.join(directory, csv_file)

        date = str.split(csv_file, '.')[0]
        
        # 读取 CSV 文件，假设文件有头部行（列名）
        df = pd.read_csv(file_path, header=None)
        
        for index, row in df.iterrows():
            sn = str(row[0])
            id = string_to_combined_hex_to_decimal(sn)
            result = int(row[col_index])

            if result == -1:
                # algo_result[id] = '-1'
                pass

            elif result != 0:
                if id not in algo_result.keys():
                    algo_result[id] = date


    return algo_result
```

### 查情标签，查情结果

```python
import pandas as pd
from math import isnan

# 读取 Excel 文件
file_path = r'E:\5.毫米波雷达\code\一步到位\excel 3.0\长塘\20250523\人工算法结果汇总-发情.xlsx'
df = pd.read_excel(file_path)

contrix = [[0, 0], [0, 0]]
manual_col = '发情日期'
algo_col = '算法发情日期'
range_day = [-1, 3]

# 按行遍历 DataFrame
for index, row in df.iterrows():
    if row['种猪状态'] == '后备':
        continue

    # 按列名取出值
    manual_date = row[manual_col]
    algo_date = row[algo_col]

    # print(algo_date, type(algo_date))

    # 都不发情
    if isnan(manual_date) and isnan(algo_date):
        contrix[1][1] = contrix[1][1] + 1

    # 人工发情，算法不发情
    elif isnan(algo_date):
        contrix[1][0] = contrix[1][0] + 1

    elif algo_date == -1:
        continue

    # 人工不发情，算法发情
    elif isnan(manual_date):
        contrix[0][1] = contrix[0][1] + 1

    else:
        diff = algo_date - manual_date

        if range_day[0] <= diff <= range_day[1]:
            contrix[0][0] = contrix[0][0] + 1
        else:
            contrix[1][0] = contrix[1][0] + 1

print(contrix)
```
