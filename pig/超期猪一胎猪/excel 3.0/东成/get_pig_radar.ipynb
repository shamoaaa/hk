```python
def string_to_combined_hex_to_decimal(input_string):
    if len(input_string) == 5:
        return input_string
        
    # 初始化空字符串用于保存十六进制表示
    combined_hex = ""

    for char in input_string:
        # 将字符转换为两位的十六进制表示并添加到combined_hex中
        combined_hex += format(ord(char), "02x")

    # 将完整的十六进制字符串转换为十进制数值
    decimal_value = int(combined_hex, 16)

    return str(decimal_value)
```

```python
# 读取 Excel 文件
file_path = r'E:\5.毫米波雷达\code\一步到位\excel 3.0\东成\查情配种记录2.xlsx'
df = pd.read_excel(file_path)

contrix = [[0, 0], [0, 0]]
manual_col = '发情日期'
algo_col = '算法发情日期'
range_day = [-1, 3]

# 按行遍历 DataFrame
for index, row in df.iterrows():
    # 按列名取出值
    manual_date = row[manual_col]
    algo_date = row[algo_col]

    # print(algo_date, type(algo_date))

    if str(row['当前胎次']) == '1':
        continue

    # 都不发情
    if isnan(manual_date) and isnan(algo_date):
        contrix[1][1] = contrix[1][1] + 1

    # 人工发情，算法不发情
    elif isnan(algo_date):
        contrix[1][0] = contrix[1][0] + 1

    elif algo_date == -1:
        continue

    # 人工不发情，算法发情
    elif isnan(manual_date):
        contrix[0][1] = contrix[0][1] + 1

    else:
        diff = algo_date - manual_date

        if range_day[0] <= diff <= range_day[1]:
            contrix[0][0] = contrix[0][0] + 1
        else:
            contrix[1][0] = contrix[1][0] + 1

print(contrix)
```
