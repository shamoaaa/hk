```python
import os

import pandas as pd
```

```python
def get_result(folder, date):

    no_pig_num = 0
    dead_pig_num = 0
    have_pig_num = 0

    for dirpath, dirnames, filenames in os.walk(folder):
        # 示例操作，例如逐个访问文件
        for filename in filenames:
            
            if date in filename:
                file_path = os.path.join(dirpath, filename)
                df = pd.read_csv(file_path, index_col=0)

                for index, row in df.iterrows():
                    alive_result = int(row.values[-1])
                    # print(alive_result)
                    if alive_result != -1:
                        if alive_result == 0:
                            no_pig_num += 1
                        elif alive_result == 2:
                            dead_pig_num += 1
                        elif alive_result == 1:
                            have_pig_num += 1

    print(no_pig_num, dead_pig_num, have_pig_num)
```

```python
folder = r'/data2/wens_data/out/0902/train_feature/changtang_20250505/'
date = '-20250507'
get_result(folder, date)
```
